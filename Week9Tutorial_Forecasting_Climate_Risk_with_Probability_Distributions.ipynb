📊 Step 2 – Binomial Distribution: Community Preparedness

Scenario: A climate response team surveys 20 communities. Each has a 65% chance of being adequately prepared for floods. Use a binomial distribution to model the number of prepared communities.

```r
n <- 20
p <- 0.65
x_binom <- 0:n
y_binom <- dbinom(x_binom, size = n, prob = p)
expected_binom <- n * p

df_binom <- data.frame(Prepared = x_binom, Probability = y_binom)

ggplot(df_binom, aes(Prepared, Probability)) +
  geom_bar(stat = "identity", fill = "steelblue") +
  geom_vline(xintercept = expected_binom, linetype = "dashed", color = "red") +
  labs(
    title = "Binomial Distribution: Community Flood Preparedness",
    subtitle = "Expected Value (dashed red line)",
    x = "Number of Prepared Communities",
    y = "Probability"
  ) +
  theme_minimal()
```

```r
#Prompt: Try changing p to 0.4 or 0.85 and observe how the distribution shifts.
n <- 20
p <- 0.4
x_binom <- 0:n
y_binom <- dbinom(x_binom, size = n, prob = p)
expected_binom <- n * p

df_binom <- data.frame(Prepared = x_binom, Probability = y_binom)

ggplot(df_binom, aes(Prepared, Probability)) +
  geom_bar(stat = "identity", fill = "steelblue") +
  geom_vline(xintercept = expected_binom, linetype = "dashed", color = "red") +
  labs(
    title = "Binomial Distribution: Community Flood Preparedness",
    subtitle = "Expected Value (dashed red line)",
    x = "Number of Prepared Communities",
    y = "Probability"
  ) +
  theme_minimal()
```

🌧 Step 3 – Poisson Distribution: Forecasting Storm Events

A coastal region averages 5 extreme storms per year. Use a Poisson distribution to model how many storms might occur next year.

```r
lambda <- 5
x_poisson <- 0:15
y_poisson <- dpois(x_poisson, lambda = lambda)

df_poisson <- data.frame(Events = x_poisson, Probability = y_poisson)

ggplot(df_poisson, aes(Events, Probability)) +
  geom_bar(stat = "identity", fill = "firebrick") +
  geom_vline(xintercept = lambda, linetype = "dashed", color = "black") +
  labs(
    title = "Poisson Distribution: Storm Frequency Forecast",
    subtitle = "Expected Number of Events = λ",
    x = "Number of Storms",
    y = "Probability"
  ) +
  theme_minimal()
```

```r
#Prompt: Try changing lambda to 3 or 7. What does this do to the shape of the distribution?
lambda <- 3
x_poisson <- 0:15
y_poisson <- dpois(x_poisson, lambda = lambda)

df_poisson <- data.frame(Events = x_poisson, Probability = y_poisson)

ggplot(df_poisson, aes(Events, Probability)) +
  geom_bar(stat = "identity", fill = "firebrick") +
  geom_vline(xintercept = lambda, linetype = "dashed", color = "black") +
  labs(
    title = "Poisson Distribution: Storm Frequency Forecast",
    subtitle = "Expected Number of Events = λ",
    x = "Number of Storms",
    y = "Probability"
  ) +
  theme_minimal()
```

🌡 Step 4 – Normal Distribution: Temperature Anomalies

```r
#Average yearly temperature anomaly in your region is 1.1°C, with a standard deviation of 0.4°C. Visualize how temperature anomalies are distributed across years.

mean_temp <- 1.1
sd_temp <- 0.4
x_temp <- seq(-0.5, 2.5, length.out = 200)
y_temp <- dnorm(x_temp, mean = mean_temp, sd = sd_temp)

df_temp <- data.frame(Anomaly = x_temp, Density = y_temp)

ggplot(df_temp, aes(Anomaly, Density)) +
  geom_line(color = "darkgreen", size = 1.2) +
  geom_vline(xintercept = mean_temp, linetype = "dashed", color = "blue") +
  labs(
    title = "Normal Distribution: Temperature Anomalies",
    subtitle = "Mean = 1.1°C (dashed line)",
    x = "Temperature Anomaly (°C)",
    y = "Density"
  ) +
  theme_minimal()
```

```r
#Prompt: Change sd_temp to 0.2 or 0.8. How does the spread of the curve change?
mean_temp <- 1.1
sd_temp <- 0.8
x_temp <- seq(-0.5, 2.5, length.out = 200)
y_temp <- dnorm(x_temp, mean = mean_temp, sd = sd_temp)

df_temp <- data.frame(Anomaly = x_temp, Density = y_temp)

ggplot(df_temp, aes(Anomaly, Density)) +
  geom_line(color = "darkgreen", size = 1.2) +
  geom_vline(xintercept = mean_temp, linetype = "dashed", color = "blue") +
  labs(
    title = "Normal Distribution: Temperature Anomalies",
    subtitle = "Mean = 1.1°C (dashed line)",
    x = "Temperature Anomaly (°C)",
    y = "Density"
  ) +
  theme_minimal()
```

🧮 Step 5 – Interpreting Expected Values

Expected values:
Prepared communities: n × p = 20 × 0.65 = 13
Storm events: λ = 5
Temperature anomalies: μ = 1.1 (mean of normal distribution)


Discussion:

How would you advise a policymaker planning flood resources based on these forecasts?

Ans: Do not plan for the average; allocate resources for extreme tail risks (e.g., preparing for 8–10 storms, not just the mean of 5) and target remaining capacity to bring the roughly 7 unprepared communities up to standard.


Which distribution shows the most variability or unpredictability?

Ans: The Poisson distribution carries the most operational unpredictability because it models unbounded, right-skewed extreme weather events where severe storm clusters can far exceed expected averages.
